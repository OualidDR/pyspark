```python
from pyspark.sql import SparkSession, functions as f
spark = (
    SparkSession.builder
    .master("local[4]")
    .appName("spark_tp")
    .config("spark.sql.shuffle.partitions", "8")
    .config("spark.driver.host", "localhost")
    .config("spark.driver.bindAddress", "127.0.0.1")
    .getOrCreate()
)
print(spark.sparkContext.uiWebUrl)
```

```python
spark.stop()
```

```python
customers = spark.read.parquet("data/customers")
orders = spark.read.parquet("data/orders")
customers.show(2)
orders.show(2)
```

```python
df = orders.join(customers, "cust_id").orderBy(F.col("amount"))
df.show(10)
```

```python
from pyspark.sql import Window

w = Window.partitionBy("country").orderBy(F.col("amount").desc())

df2 = df.withColumn("total_amount", F.sum("amount").over(w))
df2.show(100)
```
